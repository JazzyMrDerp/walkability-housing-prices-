```python
import os
import pandas as pd
from pathlib import Path

# Always resolve paths relative to this notebook's location
PROJECT_ROOT = Path(os.getcwd()).parent  # goes up from /notebooks to project root

zillow_df = pd.read_csv(PROJECT_ROOT / "data" / "zillow.csv")
zillow_df.head()
```

First lets organize it by zip code

```python
zillow_zip_sort_df = zillow_df.sort_values(["RegionName"])
zillow_zip_sort_df
```

```python
#Shape of the df
zillow_zip_sort_df.shape
```

How many NaN are there

```python
zillow_zip_sort_df.isna().sum()
```

Separate metadata from the date column

```python
meta_cols = ['RegionID', 'SizeRank', 'RegionName', 'RegionType',
             'StateName', 'State', 'City', 'Metro', 'CountyName']
print(zillow_df[meta_cols].isna().sum())
```

Drop rows missing City or Metro

```python
print(f"Rows before dropping missing City/Metro: {len(zillow_df)}")
zillow_df = zillow_df.dropna(subset=['City', 'Metro'])
print(f"Rows after: {len(zillow_df)}")
```

Grab columns from only 2021

```python
cols_2021 = [c for c in zillow_df.columns if c.startswith('2021')]
print(f"2021 columns found: {len(cols_2021)}")  # should be 12 months
print(cols_2021)
```

```python
# ── Keep metadata + 2021 columns only ─────────────────────────────────────────
zillow_2021 = zillow_df[meta_cols + cols_2021].copy()

# ── Drop rows missing City or Metro ───────────────────────────────────────────
zillow_2021 = zillow_2021.dropna(subset=['City', 'Metro'])
```

```python
zillow_2021
```

Now we'll use the average home cost in 2021

```python
zillow_2021['home_value'] = zillow_2021[cols_2021].mean(axis=1).round(2)
zillow_2021 = zillow_2021.dropna(subset=['home_value'])
```

```python
zillow_2021
```

```python
final_cols = meta_cols + ['home_value']
zillow_clean = zillow_2021[final_cols].copy()
```

```python
zillow_clean
```

Now sort by RegionName, and make that the ID

```python
zillow_clean = zillow_clean.sort_values('RegionName')
zillow_clean = zillow_clean.set_index('RegionName')
```

```python
zillow_clean = zillow_clean.rename_axis('ZIP')
zillow_clean
```
